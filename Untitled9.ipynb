## Ingest


```python
import pandas as pd

pd.options.future.infer_string = False

df = pd.read_csv("https://raw.githubusercontent.com/nytimes/covid-19-data/master/us-counties-2021.csv")
df.head()
```

```python
df["date"]=pd.to_datetime(df["date"])
df=df.set_index("date")
df.head()
```

```python
df["deaths"] = df["deaths"].fillna(
    df.groupby("county")["deaths"].transform("mean")
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

for state in df["state"].unique():
    state_data = df[df["state"] == state]

    sns.lmplot(
        data=state_data,
        x="cases",
        y="deaths",
        height=5,
        aspect=1.3
    )

    plt.title(f"Cases vs Deaths - {state}")
    plt.show()
```
